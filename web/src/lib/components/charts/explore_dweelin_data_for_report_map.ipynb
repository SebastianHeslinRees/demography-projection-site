```python
# --- 1) Prepare lookup (LSOA -> LAD) ---
lookup_cols = ["LSOA21CD", "LAD24CD", "LAD24NM"]
lookup_la = (
    latolsoa_lookup[lookup_cols]
    .dropna(subset=["LSOA21CD", "LAD24CD"])
    .drop_duplicates(subset=["LSOA21CD"])
)

# --- 2) Helper to find LSOA code column in each dataset ---
def find_lsoa_col(df: pd.DataFrame):
    candidates = [
        "LSOA21CD", "lsoa_cd", "LSOA_CD", 
    ]
    for c in candidates:
        if c in df.columns:
            return c
    return None

# --- 3) Aggregate one dataframe to LAD level ---
def aggregate_to_lad(df: pd.DataFrame, df_name: str):
    lsoa_col = find_lsoa_col(df)
    if lsoa_col is None:
        print(f"[{df_name}] No LSOA column found. Columns are:\n{list(df.columns)}")
        return None

    tmp = df.copy()
    tmp[lsoa_col] = tmp[lsoa_col].astype(str).str.strip()
    tmp = tmp.merge(lookup_la, left_on=lsoa_col, right_on="LSOA21CD", how="left")

    missing_map = tmp["LAD24CD"].isna().sum()
    if missing_map:
        print(f"[{df_name}] Warning: {missing_map} rows did not match lookup")

    # numeric columns to sum (exclude IDs if numeric)
    numeric_cols = tmp.select_dtypes(include=[np.number]).columns.tolist()
    numeric_cols = [c for c in numeric_cols if c.lower() not in {"objectid"}]

    lad_agg = (
        tmp.groupby(["LAD24CD", "LAD24NM"], dropna=False)[numeric_cols]
        .sum(min_count=1)
        .reset_index()
    )

    # optional diagnostics
    lad_agg["row_count"] = tmp.groupby(["LAD24CD", "LAD24NM"], dropna=False).size().values
    return lad_agg

# --- 4) Run for all your dataframes ---
input_dfs = {
    "large_sites_polyissues_lsoa": df_large_polyissues,
    "large_sites_withpoly_lsoa": df_large_withpoly,
    "lsoa_industrial_additionalrelease10y": df_industrial_10y,
    "phased_lsoa_longertermgb": df_phased_longerterm,
    "small_sites_lsoa": df_small_sites,
}

lad_dfs = {}
for name, d in input_dfs.items():
    if d is None:
        print(f"[{name}] DataFrame is None")
        continue
    lad_dfs[name] = aggregate_to_lad(d, name)
    if lad_dfs[name] is not None:
        print(f"[{name}] -> LAD rows: {lad_dfs[name].shape[0]}")

# Example: view one result
lad_dfs["large_sites_polyissues_lsoa"].head()
```

```python
# Build one combined LAD table from lad_dfs

# 1) Optional: normalize metric column names so they align better
def norm_col(c: str) -> str:
    return (
        c.strip()
         .lower()
         .replace("%", "pct")
         .replace(" ", "_")
         .replace("-", "_")
    )

lad_dfs_norm = {}
for ds, d in lad_dfs.items():
    if d is None or d.empty:
        continue
    d2 = d.copy()
    d2.columns = [norm_col(c) for c in d2.columns]
    lad_dfs_norm[ds] = d2

# 2) Find full metric set across datasets
id_cols = {"lad24cd", "lad24nm"}
all_metrics = sorted(
    set().union(*[
        set(df.columns) - id_cols
        for df in lad_dfs_norm.values()
    ])
)

# 3) Create one wide combined table (same columns for every dataset)
combined_lad_wide = []
for ds, d in lad_dfs_norm.items():
    out = d.copy()
    out["dataset"] = ds

    # ensure every metric exists
    for m in all_metrics:
        if m not in out.columns:
            out[m] = np.nan

    out = out[["dataset", "lad24cd", "lad24nm"] + all_metrics]
    combined_lad_wide.append(out)

combined_lad_wide = pd.concat(combined_lad_wide, ignore_index=True)

# 4) Also create a long/tidy version (best for plotting/comparison)
combined_lad_long = combined_lad_wide.melt(
    id_vars=["dataset", "lad24cd", "lad24nm"],
    value_vars=all_metrics,
    var_name="metric",
    value_name="value"
)

print("combined_lad_wide:", combined_lad_wide.shape)
print("combined_lad_long:", combined_lad_long.shape)
display(combined_lad_wide.head())
display(combined_lad_long.head())
```

```python
import re
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import ipywidgets as widgets
from IPython.display import display

# --- 1) map "year-like" column names to a numeric year (start year) ---
def extract_start_year(col_name: str):
    c = str(col_name).strip().lower()
    patterns = [
        r"^((?:19|20)\d{2})$",                          # 2024
        r"^((?:19|20)\d{2})_(?:\d{2}|\d{4})$",         # 2025_26 or 2027_2028
        r"^_((?:19|20)\d{2})_(?:\d{2}|\d{4})$",        # _2027_2028
        r"^((?:19|20)\d{2})_total$",                    # 2050_total
    ]
    for p in patterns:
        m = re.match(p, c)
        if m:
            return int(m.group(1))
    return None

year_map = {c: extract_start_year(c) for c in combined_lad_wide.columns}
year_cols = [c for c, y in year_map.items() if y is not None]

year_cols = [
    c for c, y in year_map.items()
    if y is not None and not str(c).lower().endswith("_total")
]

if not year_cols:
    raise ValueError("No year-like columns found.")

# --- 2) melt to long format and collapse duplicate year columns per LA/dataset/year ---
tmp = combined_lad_wide.melt(
    id_vars=["dataset", "lad24cd", "lad24nm"],
    value_vars=year_cols,
    var_name="year_col",
    value_name="value"
).copy()

tmp["year"] = tmp["year_col"].map(year_map)
tmp["value"] = pd.to_numeric(tmp["value"], errors="coerce")

# if multiple source columns map to same year, keep first non-null value
def first_non_null(s):
    s = s.dropna()
    return s.iloc[0] if len(s) else np.nan

plot_df = (
    tmp.groupby(["dataset", "lad24cd", "lad24nm", "year"], as_index=False)["value"]
       .agg(first_non_null)
)

# --- 3) plotting function (NaN creates line breaks) ---
def plot_dataset(ds: str, top_n_las: int = 25):
    d = plot_df[plot_df["dataset"] == ds].copy()
    if d.empty:
        print(f"No data for dataset: {ds}")
        return

    # optional clutter control
    rank = (
        d.groupby("lad24nm")["value"]
         .sum(min_count=1)
         .sort_values(ascending=False)
         .head(top_n_las)
         .index
    )
    d = d[d["lad24nm"].isin(rank)]

    fig = go.Figure()
    for la, g in d.groupby("lad24nm", dropna=False):
        g = g.sort_values("year")
        fig.add_trace(go.Scatter(
            x=g["year"],
            y=g["value"],
            mode="lines",
            name=str(la),
            connectgaps=False
        ))

    fig.update_layout(
        title=f"LA trends by year — {ds}",
        xaxis_title="Year",
        yaxis_title="Value",
        hovermode="x unified",
        template="plotly_white",
        height=700
    )
    fig.show()

# --- 4) dataset dropdown ---
dataset_dd = widgets.Dropdown(
    options=sorted(plot_df["dataset"].dropna().unique().tolist()),
    description="Dataset:"
)
topn_slider = widgets.IntSlider(value=25, min=5, max=100, step=5, description="Top LAs:")
out = widgets.Output()

def redraw(*_):
    with out:
        out.clear_output(wait=True)
        plot_dataset(dataset_dd.value, topn_slider.value)

dataset_dd.observe(redraw, names="value")
topn_slider.observe(redraw, names="value")

display(widgets.HBox([dataset_dd, topn_slider]), out)
redraw()
```

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ...existing code...

def _norm_col(c: str) -> str:
    return str(c).strip().lower().replace("%", "pct").replace(" ", "_").replace("-", "_")

def _resolve_metric_column(df_cols, candidates):
    colset = set(df_cols)
    norm_map = {_norm_col(c): c for c in df_cols}
    for c in candidates:
        if c in colset:
            return c
    for c in candidates:
        nc = _norm_col(c)
        if nc in norm_map:
            return norm_map[nc]
    return None

dataset_metric_candidates = {
    "large_sites_polyissues_lsoa": ["10Y_Total"],
    "large_sites_withpoly_lsoa": ["10Y_Total"],
    "lsoa_industrial_additionalrelease10y": ["10Y total", "10Y_Total"],
    "phased_lsoa_longertermgb": ["Sum of Adjusted10Y"],
    "small_sites_lsoa": ["10Y_Total"],
}

# 1) Lean boundary frame: only code + geometry
code_candidates = ["LAD24CD", "lad24cd", "LADCD", "ladcd", "GSS_CODE", "gss_code"]
map_code_col = next((c for c in code_candidates if c in gdf.columns), None)
if map_code_col is None:
    raise ValueError(f"No LAD code column in gdf. Available: {gdf.columns.tolist()}")

map_gdf_lean = gdf[[map_code_col, "geometry"]].copy()
map_gdf_lean[map_code_col] = map_gdf_lean[map_code_col].astype(str).str.strip().str.upper()

# 2) Lean metric frame: dataset + lad24cd + value only
lad = combined_lad_wide.copy()
lad["lad24cd"] = lad["lad24cd"].astype(str).str.strip().str.upper()

parts = []
resolved = {}
for ds, candidates in dataset_metric_candidates.items():
    metric_col = _resolve_metric_column(lad.columns, candidates)
    if metric_col is None:
        continue

    tmp = lad.loc[lad["dataset"] == ds, ["lad24cd", metric_col]].copy()
    tmp[metric_col] = pd.to_numeric(tmp[metric_col], errors="coerce")
    tmp = tmp.dropna(subset=[metric_col])
    if tmp.empty:
        continue

    tmp = tmp.rename(columns={metric_col: "value"})
    tmp["dataset"] = ds
    parts.append(tmp[["dataset", "lad24cd", "value"]])
    resolved[ds] = metric_col

if not parts:
    raise ValueError("No datasets available to plot.")

plot_values_lean = pd.concat(parts, ignore_index=True)
plot_values_lean["dataset"] = plot_values_lean["dataset"].astype("category")
plot_values_lean["lad24cd"] = plot_values_lean["lad24cd"].astype("category")
plot_values_lean["value"] = plot_values_lean["value"].astype("float32")

print("Lean values shape:", plot_values_lean.shape)
print("Resolved metrics:", resolved)

# 3) Plot from lean frames
plot_items = list(resolved.keys())
n = len(plot_items)
ncols = 2
nrows = math.ceil(n / ncols)

fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(14, 6 * nrows))
axes = np.array(axes).reshape(-1)

for i, ds in enumerate(plot_items):
    ax = axes[i]
    d = plot_values_lean.loc[plot_values_lean["dataset"] == ds, ["lad24cd", "value"]]
    m = map_gdf_lean.merge(d, left_on=map_code_col, right_on="lad24cd", how="left")

    m.plot(
        column="value",
        cmap="viridis",
        linewidth=0.6,
        edgecolor="white",
        legend=True,
        ax=ax,
        missing_kwds={"color": "lightgrey", "label": "No data"},
    )

    nn = int(m["value"].notna().sum())
    ax.set_title(f"{ds}\n(non-null LADs: {nn})", fontsize=11)
    ax.set_axis_off()

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()

# ...existing code...
```
